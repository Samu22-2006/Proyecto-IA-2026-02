```python
# 1. FUNCIÓN PARA INGRESAR LOS DATOS DEL CLIENTE

def ingresar_datos():

    cliente = {}

    cliente["nombre"] = input("Nombre del cliente: ")

    cliente["edad"] = int(
        input("Edad del cliente: ")
    )

    cliente["ingreso_mensual"] = float(
        input("Ingreso mensual: ")
    )

    cliente["monto_credito"] = float(
        input("Monto solicitado: ")
    )

    cliente["plazo"] = float(
        input("Plazo solicitado (años): ")
    )

    cliente["score"] = int(
        input("Score crediticio: ")
    )

    cliente["endeudamiento"] = float(
        input("Nivel de endeudamiento (%): ")
    )

    cliente["antiguedad_empleo_actual"] = float(
    input("Antigüedad en el empleo actual (años): ")
    )

    cliente["reportes_negativos"] = int(
        input("Número de reportes negativos: ")
    )

    return cliente

# 2. EVALUAR EL SCORE CREDITICIO

def evaluar_score(score):

    if score >= 750:
        return "BUENO"

    elif score >= 550:
        return "REGULAR"

    else:
        return "MALO"


# 3. EVALUAR EL NIVEL DE ENDEUDAMIENTO

def evaluar_endeudamiento(endeudamiento):

    if endeudamiento <= 40:
        return "BAJO"

    elif endeudamiento <= 60:
        return "MEDIO"

    else:
        return "ALTO"

# 4. EVALUAR ESTABILIDAD LABORAL

def evaluar_estabilidad_actual(antiguedad_actual, edad):
    if 17 <= edad <= 20:
        if antiguedad_actual >= 1:
            return "ESTABLE"
        else:
            return "N/A"
    elif 21 <= edad <= 24:
        if antiguedad_actual >= 1:
            return "ESTABLE"
        else:
            return "MEDIA"
    elif 25 <= edad <= 35:
        if antiguedad_actual >= 4:
            return "ESTABLE"
        elif antiguedad_actual >= 2:
            return "MEDIA"
        else:
            return "BAJA"
    elif 36 <= edad <= 50:
        if antiguedad_actual >= 6:
            return "ESTABLE"
        elif antiguedad_actual >= 4:
            return "MEDIA"
        else:
            return "BAJA"
    elif 51 <= edad <= 65:
        if antiguedad_actual >= 8:
            return "ESTABLE"
        elif antiguedad_actual >= 6:
            return "MEDIA"
        else:
            return "BAJA"
    else:
        return "N/A"

# 6. EVALUAR CAPACIDAD DE PAGO

def evaluar_capacidad_pago(plazo, ingreso, monto_credito):

    cuota = monto_credito / (plazo*12)
    relacion = cuota / ingreso

    if 0 <= relacion <= 0.2:
        return "BUENA"
    elif 0.2 < relacion <= 0.3:
        return "MEDIA"
    elif 0.3 < relacion <= 0.4:
        return "REVISAR"
    else:
        return "BAJA"

# 7. MOTOR DE INFERENCIA

def motor_inferencia(cliente):

    # El motor primero obtiene nuevos hechos
    # a partir de los datos originales

    perfil_score = evaluar_score(
        cliente["score"]
    )

    nivel_deuda = evaluar_endeudamiento(
        cliente["endeudamiento"]
    )

    estabilidad_actual= evaluar_estabilidad_actual(
    cliente["antiguedad_empleo_actual"],
    cliente["edad"]
    )

    capacidad_pago = evaluar_capacidad_pago(
        cliente["plazo"],
        cliente["ingreso_mensual"],
        cliente["monto_credito"]
    )

    # REGLAS DE RECHAZO

    if capacidad_pago == "BAJA":

        return(
            "RECHAZADO",
            "EL cliente no tiene suficiente capacidad de pago."
        )

    if cliente["reportes_negativos"] >= 2:

        return (
            "RECHAZADO",
            "El cliente tiene dos o más reportes negativos."
        )


    if perfil_score == "MALO":

        return (
            "RECHAZADO",
            "El score crediticio es demasiado bajo."
        )


    if nivel_deuda == "ALTO":

        return (
            "RECHAZADO",
            "El nivel de endeudamiento es demasiado alto."
        )

    # REGLAS DE APROBACIÓN

    if (
        perfil_score == "BUENO"
        and nivel_deuda == "BAJO"
        and (estabilidad_actual == "ESTABLE" or estabilidad_actual == "N/A")
        and (capacidad_pago == "BUENA" 
            or (capacidad_pago == "MEDIA" and nivel_deuda == "BAJO") )
        and cliente["reportes_negativos"] == 0
    ):

        return (
            "APROBADO",
            "El cliente presenta un buen perfil financiero."
        )

    # REGLAS DE REVISIÓN

    if (
        perfil_score == "REGULAR"
        or capacidad_pago == "REVISAR"
        or nivel_deuda == "MEDIO" 
        or estabilidad_actual == "MEDIA"
        or cliente["reportes_negativos"] == 1
    ):

        return (
            "REVISIÓN",
            "El cliente presenta condiciones intermedias "
            "y necesita una evaluación adicional."
        )


    # Si no cumple claramente ninguna regla anterior

    return (
        "REVISIÓN",
        "La solicitud necesita ser revisada con mayor detalle."
    )


# 8. MOSTRAR RESULTADOS

def mostrar_resultado(cliente, decision, explicacion):

    print("RESULTADO DE LA SOLICITUD")

    print("Cliente:", cliente["nombre"])

    print("Edad:", cliente["edad"])

    print("Ingresos:", cliente["ingreso_mensual"])

    print("Score:", cliente["score"])

    print(
        "Endeudamiento:",
        cliente["endeudamiento"],
        "%"
    )

    print(
        "Monto solicitado: $",
        cliente["monto_credito"]
    )

    print(
        "Plazo solicitado: ",
        cliente["plazo"]
    )

    print("\nDECISIÓN:", decision)

    print("EXPLICACIÓN:", explicacion)

# 9. PROGRAMA PRINCIPAL

cliente = ingresar_datos()

decision, explicacion = motor_inferencia(
    cliente
)

mostrar_resultado(
    cliente,
    decision,
    explicacion
)
```
